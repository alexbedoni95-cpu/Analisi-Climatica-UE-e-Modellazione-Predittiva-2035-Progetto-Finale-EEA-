### Visualizzazione aggregata del trend delle emissioni di gas serra nell'Unione Europea (1990-2023)

Di seguito, un grafico per ciascun tipo di gas che mostra il trend delle emissioni per tutti i paesi dell'Unione Europea. Questo permette di confrontare le tendenze tra i paesi per un dato gas.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

unique_gas_types = df_all_eu_gases_long['Gas_Type'].unique()

for gas_type in unique_gas_types:
    df_gas_filtered = df_all_eu_gases_long[df_all_eu_gases_long['Gas_Type'] == gas_type]

    plt.figure(figsize=(15, 8))
    sns.lineplot(data=df_gas_filtered, x='Year', y='Emission', hue='Country', marker='o', linewidth=1.5)

    plt.title(f"Trend delle emissioni di {gas_type} nell'Unione Europea (1990-2023)", fontsize=16)
    plt.xlabel("Anno", fontsize=12)
    plt.ylabel(f"Emissioni di {gas_type}", fontsize=12)
    plt.tick_params(axis='x', rotation=45)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend(title='Paese', bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.2, ncol=2)
    plt.tight_layout(rect=[0, 0, 0.88, 1]) # Adjust layout to prevent legend overlap
    plt.show()
```
